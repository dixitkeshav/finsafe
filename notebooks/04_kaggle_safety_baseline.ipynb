# FinSafe 04 — freeze a safety pilot and collect the original-model baseline
Run after 03 in the same session. This does not start full training. The pilot has 12 validation behavior groups, each with a direct request, one fixed attack and a benign prevention question. It is AI-assisted and not independently validated. Test prompts remain reserved. Review responses before reporting safety rates. Every cell logs time and failure; generation resumes per response. Download the recovery archive before ending the session.

```python
import sys, os, json, subprocess, time, datetime
from pathlib import Path
started=time.monotonic()
print(f"[{datetime.datetime.now(datetime.timezone.utc).isoformat()}] START stage04 bootstrap",flush=True)
try:
    project=Path("/kaggle/working/finsafe")
    if not project.exists():
        raise RuntimeError("Restore the project/data or run notebook 01 and 02 first")
    status=subprocess.check_output(["git","-C",str(project),"status","--porcelain","--untracked-files=no"],text=True)
    if status.strip():
        raise RuntimeError("Save tracked-file edits before updating")
    subprocess.run(["git","-C",str(project),"pull","--ff-only"],check=True)
    sys.path.insert(0,str(project/"src"))
    from finsafe.progress import Progress
    progress=Progress(project/"runs/notebook-04/logs","notebook-04")
    progress.log(f"DONE bootstrap in {time.monotonic()-started:.1f}s")
except BaseException as error:
    print(f"FAILED bootstrap after {time.monotonic()-started:.1f}s ({type(error).__name__})",flush=True)
    raise
```

```python
with progress.stage("01-prerequisites"):
    from kaggle_secrets import UserSecretsClient
    from finsafe.data import read_jsonl
    runtime_env=os.environ.copy()
    runtime_env.update(HF_TOKEN=UserSecretsClient().get_secret("HF_TOKEN"),
        HF_HOME="/kaggle/working/.cache/huggingface",HF_NAMESPACE="Finsafe-team",
        CUDA_VISIBLE_DEVICES="0",TOKENIZERS_PARALLELISM="false",TRL_ONLY_MODE="1",
        DISABLE_UNSLOTH_FOR_TRL="1",PYTHONUNBUFFERED="1")
    gpu_py=str(project/".venv-gpu/bin/python")
    cli=str(project/".venv-gpu/bin/finsafe")
    revisions=json.loads((project/"runs/environment/model-revisions.json").read_text())
    global_sha=revisions["CohereLabs/tiny-aya-global"]
    filtered=project/"data/processed/tatqa-tokenized"
    curated=filtered/"curated"
    train_rows=read_jsonl(curated/"sft_messages.jsonl")
    if len(train_rows)<40:
        raise RuntimeError("Too few accepted training examples for the 5-step, effective-batch-8 smoke test")
    progress.log(f"Accepted curated training examples: {len(train_rows)}")
    progress.log((filtered/"token_manifest.json").read_text())
    progress.run([cli,"doctor","--gpu"],cwd=project,env=runtime_env)
```

```python
with progress.stage("02-freeze-safety-pilot"):
    progress.run([gpu_py,"scripts/prepare_safety_pilot.py"],cwd=project,env=runtime_env)
    progress.log((project/"data/processed/safety-pilot-v1/manifest.json").read_text())
```

## Diagnose finance output format without changing accuracy
Re-score already saved predictions. This does not regenerate the original model or change the strict answer-and-scale rule. Format errors are distinct from empty/incoherent outputs. The old generation file lacks a per-row cap field; the diagnostic below uses its known 64-token cap.

```python
with progress.stage("03-finance-baseline-diagnostics"):
    saved=project/"runs/baseline-validation-4bit/predictions.jsonl"
    progress.run([cli,"summarize","--data",str(filtered/"validation.jsonl"),"--predictions",str(saved),"--out","runs/baseline-validation-4bit/report-with-format-diagnostics.json"],cwd=project,env=runtime_env)
    predictions=read_jsonl(saved)
    progress.log(f"Original finance baseline: {len(predictions)} responses, {sum(p['output_tokens']>=64 for p in predictions)} at the known 64-token cap")
```

```python
with progress.stage("04-safety-validation-baseline"):
    safety_data=project/"data/processed/safety-pilot-v1/validation.jsonl"
    safety_predictions=project/"runs/safety-baseline-pilot-v1/predictions.jsonl"
    progress.run([gpu_py,"-u","scripts/generate.py","--model","CohereLabs/tiny-aya-global",
        "--revision",global_sha,"--data",str(safety_data),"--condition","aligned-safety-pilot-v1",
        "--out",str(safety_predictions),"--max-new-tokens","256","--load-in-4bit","--resume"],cwd=project,env=runtime_env)
```

```python
with progress.stage("05-review-packet"):
    from finsafe.progress import atomic_json
    review_packet=project/"runs/safety-baseline-pilot-v1/review.jsonl"
    progress.run([gpu_py,"scripts/review_packet.py","--data",str(safety_data),"--predictions",str(safety_predictions),"--out",str(review_packet)],cwd=project,env=runtime_env)
    predictions=read_jsonl(safety_predictions)
    capped=sum(p["output_tokens"]>=256 for p in predictions)
    atomic_json(project/"runs/safety-baseline-pilot-v1/status.json",{"n":len(predictions),"cap_hits":capped,"cap":256,"safety_rates":None,"label_status":"UNREVIEWED"})
    progress.log(f"Safety baseline saved: {len(predictions)} responses; {capped} hit the output cap. Inspect capped responses before labeling.")
```

```python
with progress.stage("06-review-and-recovery-backup"):
    import zipfile
    from IPython.display import FileLink,display
    backup=Path("/kaggle/working/finsafe-stage04-recovery.zip")
    files=[f for folder in [project/"data/processed",project/"runs",project/"checkpoints/smoke-sft-five"] for f in folder.rglob("*") if f.is_file() and not f.is_symlink() and not f.name.startswith(".env")]
    progress.log(f"Backing up {len(files)} files ({sum(f.stat().st_size for f in files)/2**20:.1f} MiB)")
    with zipfile.ZipFile(backup,"w",compression=zipfile.ZIP_STORED) as archive:
        for i,file in enumerate(files,1):
            archive.write(file,str(file.relative_to(project)))
            if i%25==0 or i==len(files):progress.log(f"BACKUP {i}/{len(files)}")
    with zipfile.ZipFile(backup) as archive:
        if archive.testzip():raise RuntimeError("Recovery ZIP failed integrity check")
    display(FileLink(str(backup)))
    display(FileLink(str(review_packet)))
    progress.log("Stage 04 complete. Safety responses require review; no safety rates are claimed.")
```
